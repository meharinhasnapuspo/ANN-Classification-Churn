```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns ,warnings
warnings.filterwarnings("ignore")
import pickle 
import tensorflow as tf
from tensorflow.keras.models import load_model
```

```python
## load the ann model 
model = load_model('model.h5')

## load scall encode file
with open("gender_encode.pkl","rb") as file:
    gender_encode = pickle.load(file)

with open("geography_encode.pkl","rb") as file:
    geography_encode  = pickle.load(file)

with open("scale.pkl","rb") as file:
    scale  = pickle.load(file)
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encode = geography_encode.transform([[input_data['Geography']]]).toarray()
geo_enocode_df = pd.DataFrame(geo_encode,columns=geography_encode.get_feature_names_out(["Geography"]))
geo_enocode_df
```

```python
## convert datafeam
input_data = pd.DataFrame([input_data])
input_data
```

```python
## combine onh data to input data
input_data = pd.concat([input_data.drop("Geography",axis=1).reset_index(drop=True),geo_enocode_df],axis=1)
input_data
```

```python
## encode gender columns
input_data['Gender']= gender_encode.transform(input_data['Gender'])
```

```python
input_data
```

```python
## scal data
input_data = scale.transform(input_data)
input_data
```

```python
## prediction
pred = model.predict(input_data)
print(pred)
```

```python
if pred > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
